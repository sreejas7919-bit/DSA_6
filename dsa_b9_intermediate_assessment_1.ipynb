## Challenge 1: Word Frequency Analyzer
Write a function `top_k_words(filename: str, k: int) -> list` that returns the top k frequent words.
- Ignore case and punctuation.
- Sort by frequency, then alphabetically.


```python
import re
from collections import Counter


def top_k_words(filename: str, k: int) -> list:
  try:
    with open(filename, "r") as f:
      text = f.read().lower()

    words = re.findall(r"\b\w+\b", text)
    counts = Counter(words)
    sorted_words = sorted(counts.keys(), key=lambda w: (-counts[w], w))

    return sorted_words[:k]
  except FileNotFoundError:
    return []


# Run & Test
with open("sample.txt", "w") as f:
  f.write(
      "Data science is evolving. Python is versatile. Learning never stops."
  )

print(top_k_words("sample.txt", 3))
```

## Challenge 2 : Analyze daily expenses and extract insights.
User enters:

Number of transactions
Category and amount

Food 200
Travel 500
Food 300
Shopping 700

### Output
* Total expense
* Highest expense category
* Category-wise total

```python
def analyze_expenses():
  n = int(input("Enter number of transactions: "))
  expenses = {}

  for _ in range(n):
    category, amount = input().split()
    amount = float(amount)
    expenses[category] = expenses.get(category, 0) + amount

  total = sum(expenses.values())
  highest_category = max(expenses, key=expenses.get)

  print("Total Expense:", total)
  print("Highest Category:", highest_category)
  print("Category-wise:", expenses)


analyze_expenses()
```

### Challenge 3: Data Cleaner (Mini Pandas Logic Without Pandas)

10 ,20 ,abc ,30 ,-5 , 40

* Remove invalid values
* Remove negative values


Show:
* Clean list
* Average
* Max value

```python
def clean_data():
  n = int(input("Enter number of values: "))
  data = []

  for _ in range(n):
    val = input()
    try:
      num = float(val)
      if num >= 0:
        data.append(num)
    except ValueError:
      pass

  avg = sum(data) / len(data) if data else 0.0
  max_val = max(data) if data else None

  print("Clean Data:", data)
  print("Average:", avg)
  print("Max:", max_val)


clean_data()
```

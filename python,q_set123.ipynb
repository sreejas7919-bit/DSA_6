```python
#1. Write a Python program that takes an integer as input and calculates the sum of its digits
num = int(input("Enter an integer: "))
sum_val = 0
for d in str(abs(num)):
    sum_val += int(d)
print("Sum of digits:", sum_val)
```

```python
#2. Create a Python program to calculate the factorial of a given number using a function.
def factorial(n):
    fact = 1
    for i in range(1, n + 1):
        fact = fact * i
    return fact

num = int(input("Enter a number: "))
print("Factorial:", factorial(num))
```

```python
#3. Develop a Python program that checks whether a given word or phrase is a palindrome.
text = input("Enter a word or phrase: ")
text = text.lower().replace(" ", "")
if text == text[::-1]:
    print("It is a palindrome.")
else:
    print("It is not a palindrome.")
```

```python
# 4. Count vowels and consonants[cite: 1]
sentence = input("Enter a sentence: ")
vowels = "aeiouAEIOU"
v_count = 0
c_count = 0

for ch in sentence:
    if ch.isalpha():
        if ch in vowels:
            v_count += 1
        else:
            c_count += 1

print("Vowels:", v_count)
print("Consonants:", c_count)
```

```python
# 5. Common elements between two lists[cite: 1]
list1 = [1, 2, 3, 4, 5]
list2 = [3, 4, 5, 6, 7]
common = []

for x in list1:
    if x in list2 and x not in common:
        common.append(x)

print("Common elements:", common)
```

```python
# 6. Frequency of each character[cite: 1]
text = input("Enter a string: ")
counts = {}

for ch in text:
    if ch in counts:
        counts[ch] += 1
    else:
        counts[ch] = 1

print("Character frequency:", counts)
```

```python
# 7. Leap year check[cite: 1]
year = int(input("Enter a year: "))
if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
    print(year, "is a leap year.")
else:
    print(year, "is not a leap year.")
```

```python
# 8. Prime number check[cite: 1]
num = int(input("Enter a number: "))
is_prime = True

if num <= 1:
    is_prime = False
else:
    for i in range(2, num):
        if num % i == 0:
            is_prime = False
            break

if is_prime:
    print(num, "is a prime number.")
else:
    print(num, "is not a prime number.")
```

```python
# 9. Fibonacci sequence[cite: 1]
n_terms = int(input("Enter number of terms: "))
a, b = 0, 1

for _ in range(n_terms):
    print(a, end=" ")
    a, b = b, a + b
print()
```

```python
# 10. Roots of quadratic equation[cite: 1]
a = float(input("Enter a: "))
b = float(input("Enter b: "))
c = float(input("Enter c: "))

d = (b**2) - (4 * a * c)
root1 = (-b + d**0.5) / (2 * a)
root2 = (-b - d**0.5) / (2 * a)

print("Roots are:", root1, "and", root2)
```

```python
# 1. Character count[cite: 2]
text = "assembly"
counts = {}
for ch in text:
    counts[ch] = counts.get(ch, 0) + 1

res = []
for k, v in counts.items():
    res.append(f"{k}={v}")
print(", ".join(res))
```

```python
# 2. Max of three numbers[cite: 2]
def find_max(a, b, c):
    if a >= b and a >= c:
        return a
    elif b >= a and b >= c:
        return b
    else:
        return c

print(find_max(34, 12, 7))
```

```python
# 3. Exponent function[cite: 2]
def exponent(base, exp):
    return base ** exp

print("Output:", exponent(2, 3))
```

```python
# 4. Sum of cube of smaller positive integers[cite: 2]
def sum_cubes(n):
    total = 0
    for i in range(1, n):
        total += i**3
    return total

print("Output:", sum_cubes(4))
```

```python
# 5. FizzBuzz (1 to 10)[cite: 2]
for i in range(1, 11):
    if i % 2 == 0 and i % 5 == 0:
        print("FizzBuzz")
    elif i % 2 == 0:
        print("Fizz")
    elif i % 5 == 0:
        print("Buzz")
    else:
        print(i)
```

```python
# 6. Most frequent item[cite: 2]
nums = [2, 3, 4, 2, 5, 2]
most_freq = nums[0]
max_count = 0

for x in nums:
    if nums.count(x) > max_count:
        max_count = nums.count(x)
        most_freq = x

print("Most frequent item:", most_freq)
```

```python
# 7. Sum of squares[cite: 2]
nums = [2, 1, 3, 1]
total = 0
for x in nums:
    total += x**2
print("Sum of squares:", total)
```

```python
# 8. Odd or even (1 to 15)[cite: 2]
for i in range(1, 16):
    if i % 2 == 0:
        print(f"{i}-even")
    else:
        print(f"{i}-odd")
```

```python
# 9. Fahrenheit to Celsius[cite: 2]
f = 41
c = (f - 32) * 5 / 9
print("Temperature in Celsius =", int(c))
```

```python
# 10. Factorial function[cite: 2]
def factorial(n):
    if n == 0 or n == 1:
        return 1
    return n * factorial(n - 1)

print(factorial(3))
```

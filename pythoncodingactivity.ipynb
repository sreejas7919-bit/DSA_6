```python
# 1. Output of print(type(5 + 5.0))
print(type(5 + 5.0))

# 2. Result of str(5) + " apples"
print(str(5) + " apples")

# 3. Length of "Python Rocks!"
print(len("Python Rocks!"))

# 4. Add, subtract, multiply, and divide 15 and 4
print("Add:", 15 + 4)
print("Subtract:", 15 - 4)
print("Multiply:", 15 * 4)
print("Divide:", 15 / 4)

# 5. Floor division of 15 by 4
print(15 // 4)

# 6. Remainder when 17 is divided by 5
print(17 % 5)

# 7. 3 to the power of 4
print(3 ** 4)

# 8. Evaluate 10 + 2 * 5
# Explanation: Multiplication (*) has higher precedence than addition (+), so 2*5 = 10, then 10+10 = 20.
print(10 + 2 * 5)

# 9. Result of 7 % 3 == 1
print(7 % 3 == 1)

# 10. Store 3 ** 2 + 1 in a variable and print its type
result = 3 ** 2 + 1
print(type(result))

# 11. Check if 8 // 3 equals 2
print(8 // 3 == 2)

# 12. Show if len("car") == 3
print(len("car") == 3)

# 13. Check if "Python" != "python"
print("Python" != "python")

# 14. Compare the type of "10" with 10
print(type("10") == type(10))

# 15. Check if None == False
print(None == False)

# 16. Show whether 10 >= 10
print(10 >= 10)

# 17. Check if a number is between 10 and 20
num = 15
print(10 < num < 20)

# 18. Check if length of "1234567890" is even
print(len("1234567890") % 2 == 0)

# 19. Check if string contains "@" and "."
email = "user@example.com"
print("@" in email and "." in email)

# 20. Check whether the type of len("hello") is int
print(type(len("hello")) == int)
```

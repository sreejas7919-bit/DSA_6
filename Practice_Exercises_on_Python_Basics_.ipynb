```python
# 7. Reverse list using for loop
lst = [10, 40, 30, 70]
rev_lst = []
for i in range(len(lst) - 1, -1, -1):
  rev_lst.append(str(lst[i]))
print(",".join(rev_lst))
```

Practice Problems – Python Basics (Set 2)

```python
# 1. Sum of digits of an integer
num = int(input("Enter an integer: "))
total = sum(int(digit) for digit in str(abs(num)))
print("Sum of digits:", total)
```

```python
# 2. Factorial using a function
def factorial(n):
  fact = 1
  for i in range(1, n + 1):
    fact *= i
  return fact


num = int(input("Enter a number: "))
print("Factorial:", factorial(num))
```

```python
# 3. Palindrome check
text = input("Enter a word or phrase: ")
clean_text = "".join(ch.lower() for ch in text if ch.isalnum())
if clean_text == clean_text[::-1]:
  print("It is a palindrome.")
else:
  print("It is not a palindrome.")
```

```python
# 4. Count vowels and consonants
sentence = input("Enter a sentence: ")
vowels = "aeiouAEIOU"
v_count = 0
c_count = 0
for ch in sentence:
  if ch.isalpha():
    if ch in vowels:
      v_count += 1
    else:
      c_count += 1
print(f"Vowels: {v_count}, Consonants: {c_count}")
```

```python
# 5. Common elements between two lists
list1 = [1, 2, 3, 4, 5]
list2 = [3, 4, 5, 6, 7]
common = list(set(list1) & set(list2))
print("Common elements:", common)
```

```python
# 6. Character frequency in a string
text = input("Enter a string: ")
freq = {}
for ch in text:
  freq[ch] = freq.get(ch, 0) + 1
print("Character frequency:", freq)
```

```python
# 7. Leap year check
year = int(input("Enter a year: "))
if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
  print(f"{year} is a leap year.")
else:
  print(f"{year} is not a leap year.")
```

```python
# 8. Prime number check
num = int(input("Enter a number: "))
if num > 1:
  for i in range(2, int(num**0.5) + 1):
    if num % i == 0:
      print(f"{num} is not a prime number.")
      break
  else:
    print(f"{num} is a prime number.")
else:
  print(f"{num} is not a prime number.")
```

```python
# 9. Fibonacci sequence
terms = int(input("Enter number of terms: "))
a, b = 0, 1
fib_seq = []
for _ in range(terms):
  fib_seq.append(a)
  a, b = b, a + b
print("Fibonacci sequence:", fib_seq)
```

```python
# 10. Roots of a quadratic equation
a = float(input("Enter coefficient a: "))
b = float(input("Enter coefficient b: "))
c = float(input("Enter coefficient c: "))

d = (b**2) - (4 * a * c)
root1 = (-b + (d**0.5)) / (2 * a)
root2 = (-b - (d**0.5)) / (2 * a)

print(f"Roots are: {root1} and {root2}")
```

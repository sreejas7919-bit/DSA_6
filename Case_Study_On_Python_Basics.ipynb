#1. Bank Account Manager

```python
account_holder = "Sreethi"
account_number = 123456789
balance = 1000.0
```

```python
def check_balance():
    print(f"Account Holder: {account_holder}")
    print(f"Account Number: {account_number}")
    print(f"Current Balance: ₹{balance}\n")
check_balance()
```

```python
def deposit(amount):
    global balance
    if amount > 0:
        balance += amount
        print(f"Deposited: ₹{amount}")
        print(f"New Balance: ₹{balance}\n")
    else:
        print("Invalid deposit amount.\n")
```

```python
def withdraw(amount):
    global balance
    if amount > balance:
        print(f"Insufficient balance! Cannot withdraw ₹{amount}.")
        print(f"Current Balance: ₹{balance}\n")
    elif amount <= 0:
        print("Invalid withdrawal amount.\n")
    else:
        balance -= amount
        print(f"Withdrawn: ₹{amount}")
        print(f"Remaining Balance: ₹{balance}\n")
```

```python
print("BANK ACCOUNT MANAGER ")

# Check  balance
check_balance()

# Deposit
deposit(500)

# Withdrawal with sufficient balance
withdraw(200)

# Withdrawal with insufficient balance
withdraw(2000)
```

# 2. Weather Data Analyzer

a. Analyze weather data (temperature, humidity, etc.) for a given region
b. Use variables and data types to store weather data
c. Implement conditional statements to determine weather conditions
(sunny, rainy, etc.)
d. Use functions to calculate average temperature and humidity, and print
results

```python
temperatures = [30.5, 32.0, 28.0, 26.5, 31.0, 29.5, 27.0]  # in Celsius
humidity = [75, 60, 85, 90, 65, 70, 88]
```

```python
def get_avg_temperature(temp_list):
    return sum(temp_list) / len(temp_list)
```

```python
def get_avg_humidity(humidity_list):
    return sum(humidity_list) / len(humidity_list)
```

```python
def get_weather_condition(temp, hum):
    if hum >= 80:
        return "Rainy / Overcast"
    elif temp >= 30 and hum < 70:
        return "Sunny"
    else:
        return "Pleasant"
```

```python
print(" WEATHER DATA ANALYZER")

avg_temp = get_avg_temperature(temperatures)
avg_hum = get_avg_humidity(humidity)

print(f"Average Temperature: {avg_temp:.2f}°C")
print(f"Average Humidity: {avg_hum:.2f}%\n")

print("Daily Weather Breakdown:")
for i in range(len(temperatures)):
    day_temp = temperatures[i]
    day_hum = humidity[i]
    condition = get_weather_condition(day_temp, day_hum)
    print(f"Day {i+1}: {day_temp}°C, {day_hum}% Humidity -> {condition}")
```

```python
# Module 5
# Task 1: Get product price function
def get_product_price(product_id, products_list):
    for p in products_list:
        if p["Product_ID"] == product_id:
            return p["Price"]
    return None

# Task 2: Get customer name function
def get_customer_name(customer_id, customers_list):
    for c in customers_list:
        if c["Customer_ID"] == customer_id:
            return c["Name"]
    return None

# Task 3: Calculate order value function
def calculate_order_value(price, quantity):
    return price * quantity

# Task 4: Order classification with default arguments
def classify_order(value, high_limit=3000, low_limit=1000):
    if value > high_limit:
        return "High"
    elif value >= low_limit:
        return "Medium"
    else:
        return "Low"

# Task 5: Formatted invoice generator
def generate_invoice(order, products_list, customers_list):
    price = get_product_price(order["Product_ID"], products_list)
    cust_name = get_customer_name(order["Customer_ID"], customers_list)
    order_val = calculate_order_value(price, order["Quantity"])

    invoice = (
        f"INVOICE for Order {order['Order_ID']}\n"
        f"Customer Name : {cust_name}\n"
        f"Item Price    : ₹{price}\n"
        f"Quantity      : {order['Quantity']}\n"
        f"Total Amount  : ₹{order_val}\n"
        f"Status        : {order['Status']}\n"
    )
    return invoice

# Task 6: Function to calculate total revenue
def calculate_total_revenue(orders_list, products_list):
    total = 0
    for o in orders_list:
        if o["Status"] == "Cancelled":
            continue
        price = get_product_price(o["Product_ID"], products_list)
        total += calculate_order_value(price, o["Quantity"])
    return total

# Task 7: Function to identify top customer by spend (Delivered orders only)
def top_customer_by_spend(orders_list, products_list, customers_list):
    customer_spend = {}
    for o in orders_list:
        if o["Status"] == "Delivered":
            price = get_product_price(o["Product_ID"], products_list)
            val = calculate_order_value(price, o["Quantity"])
            cid = o["Customer_ID"]
            customer_spend[cid] = customer_spend.get(cid, 0) + val

    top_cid = max(customer_spend, key=customer_spend.get)
    return get_customer_name(top_cid, customers_list)

# Task 8: Bonus (*args) discount application
def apply_offer(price, *discounts):
    final_price = price
    for d in discounts:
        final_price -= final_price * (d / 100)
    return final_price

# --- Test Functions ---
print("--- Module 5 Function Tests ---")
print(generate_invoice(orders[0], products, customers))

mod5_rev = calculate_total_revenue(orders, products)
print(f"Total Revenue via Function: ₹{mod5_rev}")

top_cust = top_customer_by_spend(orders, products, customers)
print(f"Top Customer by Spend: {top_cust}")

discounted = apply_offer(1000, 10, 5)
print(f"Discounted Price (₹1000 with 10% then 5% off): ₹{discounted:.2f}")
```

```python
# Module 6
def generate_capstone_summary(products_list, customers_list, orders_list):
    # 1. Product details
    total_products = len(products_list)
    categories = set(p["Category"] for p in products_list)
    total_categories = len(categories)
    out_of_stock_count = sum(1 for p in products_list if p["Stock"] == 0)

    # 2. Customer details
    total_customers = len(customers_list)
    premium_cnt = sum(1 for c in customers_list if c["Membership"] == "Premium")
    regular_cnt = sum(1 for c in customers_list if c["Membership"] == "Regular")

    # 3. Order status counts
    total_orders = len(orders_list)
    delivered_cnt = sum(1 for o in orders_list if o["Status"] == "Delivered")
    pending_cnt = sum(1 for o in orders_list if o["Status"] == "Pending")
    cancelled_cnt = sum(1 for o in orders_list if o["Status"] == "Cancelled")

    # 4. Revenue metrics (Delivered orders only)
    delivered_revenue = 0
    delivered_categories = {}

    for o in orders_list:
        if o["Status"] == "Delivered":
            price = get_product_price(o["Product_ID"], products_list)
            val = calculate_order_value(price, o["Quantity"])
            delivered_revenue += val

            # Identify category for popularity tracking
            for p in products_list:
                if p["Product_ID"] == o["Product_ID"]:
                    cat = p["Category"]
                    delivered_categories[cat] = delivered_categories.get(cat, 0) + o["Quantity"]
                    break

    avg_order_val = delivered_revenue / delivered_cnt if delivered_cnt > 0 else 0
    top_customer = top_customer_by_spend(orders_list, products_list, customers_list)
    most_popular_category = max(delivered_categories, key=delivered_categories.get) if delivered_categories else "N/A"

    # Printing Consolidated Report
    print("=" * 60)
    print("        URBANCART BUSINESS SUMMARY REPORT        ")
    print("=" * 60)
    print(f"Total Products Catalogued : {total_products}")
    print(f"Distinct Product Categories: {total_categories}")
    print(f"Out of Stock Products     : {out_of_stock_count}")
    print("-" * 60)
    print(f"Total Customer Base       : {total_customers} (Premium: {premium_cnt} | Regular: {regular_cnt})")
    print("-" * 60)
    print(f"Total Orders Processed    : {total_orders}")
    print(f"Order Status Breakdown    : Delivered: {delivered_cnt} | Pending: {pending_cnt} | Cancelled: {cancelled_cnt}")
    print("-" * 60)
    print(f"Total Revenue (Delivered) : ₹{delivered_revenue:.2f}")
    print(f"Average Order Value (AOV) : ₹{avg_order_val:.2f}")
    print(f"Top Customer by Spend     : {top_customer}")
    print(f"Most Popular Category     : {most_popular_category}")
    print("=" * 60)

# Execute Capstone Summary Report
generate_capstone_summary(products, customers, orders)
```
